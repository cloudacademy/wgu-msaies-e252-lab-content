```python
section("nvidia-smi Output")

try:
    result = subprocess.run(
        ["nvidia-smi"],
        capture_output=True,
        text=True,
        check=True
    )

    print(result.stdout)

except FileNotFoundError:
    print("nvidia-smi not found")

except subprocess.CalledProcessError as e:
    print(f"nvidia-smi failed: {e}")
    print(e.stderr)
```

```python
section("PyTorch Check")

try:
    import torch

    print(f"PyTorch Version: {torch.__version__}")
    print(f"CUDA Available: {torch.cuda.is_available()}")

    if not torch.cuda.is_available():
        print("No CUDA GPUs detected by PyTorch")
    else:
        gpu_count = torch.cuda.device_count()

        print(f"PyTorch detected {gpu_count} GPU(s)")

        for idx in range(gpu_count):
            print(f"\nGPU {idx}")
            print(f"Name: {torch.cuda.get_device_name(idx)}")

            props = torch.cuda.get_device_properties(idx)

            print(f"Memory: {props.total_memory / (1024**3):.2f} GB")
            print(f"Compute Capability: {props.major}.{props.minor}")

except ImportError:
    print("PyTorch is not installed")

except Exception as e:
    print(f"PyTorch error: {e}")
```

```python
section("PyTorch GPU Computation")

try:
    import torch

    print(f"PyTorch Version: {torch.__version__}")

    print("Checking CUDA availability...")
    print(torch.cuda.is_available())

    if torch.cuda.is_available():

        print("Getting device count...")
        print(torch.cuda.device_count())

        print("Creating CUDA device...")
        device = torch.device("cuda:0")

        print("Allocating tensor...")
        a = torch.tensor([1.0, 2.0, 3.0], device=device)

        print("Tensor allocated:")
        print(a)

        print("Running simple addition...")
        b = a + 1

        print("Result:")
        print(b)

        print("Success!")

except Exception as e:
    print(f"PyTorch GPU test failed: {e}")
```
